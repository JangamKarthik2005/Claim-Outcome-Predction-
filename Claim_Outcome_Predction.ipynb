```python
import pandas as pd
import numpy as np
import random


random.seed(42)
np.random.seed(42)

total_claims = 400


statuses = (
    ["Paid"] * 280 +
    ["Partially_Paid"] * 80 +
    ["Denied"] * 40
)

random.shuffle(statuses)

# Sa
payers = ["Aetna", "Cigna", "UnitedHC", "BlueCross"]
providers = ["Dr.Rao", "Dr.Shah", "Dr.Lee", "Dr.Kumar"]
cpt_codes = ["99213", "99214", "99215", "99203"]
diagnosis_codes = ["M54.5", "J06.9", "E11.9", "I10"]

data = []

for i in range(total_claims):

    status = statuses[i]


    if status == "Paid":
        amount = np.random.uniform(100, 250)

    elif status == "Partially_Paid":
        amount = np.random.uniform(200, 400)

    else:
        amount = np.random.uniform(300, 600)

    claim = {
        "id": i + 1,
        "payer": random.choice(payers),
        "provider": random.choice(providers),
        "cpt_code": random.choice(cpt_codes),
        "diagnosis_code": random.choice(diagnosis_codes),
        "billed_amount": round(amount, 2),
        "patient_age": random.randint(18, 80),
        "status": status
    }

    data.append(claim)


df = pd.DataFrame(data)


df.to_csv("claims.csv", index=False)

print("Dataset created successfully!")
print("Total records:", len(df))


display(df.head())


print("\nClaim Outcome Distribution:")
print(df["status"].value_counts())
```

```python


print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```

```python
# Step 2: Exploratory Data Analysis


print("Duplicate Records:")
print(df.duplicated().sum())

print("\nClaim Status Counts:")
print(df['status'].value_counts())


print("\nNumerical Data Summary:")
print(df.describe())


print("\nUnique Payers:")
print(df['payer'].nunique())

print("\nUnique Providers:")
print(df['provider'].nunique())


print("\nFirst 10 Records:")
display(df.head(10))
```

```python
df = df.drop('id', axis=1)


X = df.drop('status', axis=1)
y = df['status']


X = pd.get_dummies(X, drop_first=True)

print("\nTarget Values:")
print(y.head())

print("\nProcessed Dataset Shape:")
print(X.shape)
```

```python


from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts())

print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python


from sklearn.ensemble import RandomForestClassifier


model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)


model.fit(X_train, y_train)

print("Model Training Completed Successfully!")


y_pred = model.predict(X_test)

print("\nFirst 10 Predicted Outcomes:")
print(y_pred[:10])
```
